# Netflix Catalogue Analysis

## 1. Introduction
Netflix’s catalogue reflects both its global expansion and its focus on different genres.  
This project answers two questions:  
1. Which countries contribute the most titles, and how has that changed over time?  
2. What genres dominate, and do they differ for Movies vs TV Shows?  

By exploring these, we gain insight into Netflix’s geographic diversity and genre focus.

```python
# Fill missing values
df['country'] = df['country'].fillna("Unknown")
df['director'] = df['director'].fillna("Unknown")

# Convert date_added to datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['year_added'] = df['date_added'].dt.year

# Split genres into lists
df['listed_in'] = df['listed_in'].str.split(', ')
```

```python
# Count titles per country per year
country_year = df.groupby(['country','year_added']).size().reset_index(name='count')

# Top 10 countries overall
top_countries = df['country'].value_counts().head(10).index
filtered = country_year[country_year['country'].isin(top_countries)]

# Pivot for plotting
pivot_country = filtered.pivot(index='year_added', columns='country', values='count').fillna(0)

# Plot
pivot_country.plot(figsize=(12,6))
plt.title("Top Countries Contributing Titles Over Time")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")
plt.legend(title="Country")
plt.show()
```

```python
# Expand genres
genres = df.explode('listed_in')

# Count by type and genre
genre_counts = genres.groupby(['type','listed_in']).size().reset_index(name='count')

# Pivot for comparison
pivot_genre = genre_counts.pivot(index='listed_in', columns='type', values='count').fillna(0)

# Top 15 genres by Movies
pivot_genre = pivot_genre.sort_values(by='Movie', ascending=False).head(15)

# Plot
pivot_genre.plot(kind='bar', figsize=(12,6))
plt.title("Top Genres in Netflix Catalogue: Movies vs TV Shows")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.legend(title="Type")
plt.show()
```

**Findings:**
1. The **United States contributes the most titles**, followed by **India and the UK**. India’s contribution has grown steadily, showing Netflix’s global expansion.  
2. **Movies dominate Drama, Comedy, and Action**, while **TV Shows dominate Kids/Family and Reality genres**.  
3. Documentaries and International content are strong in both categories, reflecting Netflix’s diverse catalogue.

**Conclusion:**  
Netflix’s catalogue shows two clear strategies:  
- **Geographic expansion**: The U.S. leads, but India and the UK are rising contributors.  
- **Genre diversification**: Movies lean toward traditional storytelling genres, while TV Shows emphasize serialized formats for ongoing engagement.  

Together, these trends highlight Netflix’s global reach and varied content strategy.
